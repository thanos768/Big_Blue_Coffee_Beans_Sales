```python
import pandas as pd
import numpy as np
```

```python
df_ord=pd.read_excel('Raw Data.xlsx',sheet_name='orders')
df_ord=df_ord[['Order ID','Order Date','Customer ID','Product ID','Quantity']]
df_cust = pd.read_excel('Raw Data.xlsx',sheet_name = 'customers')
df_prod = pd.read_excel('Raw Data.xlsx',sheet_name = 'products')
```

```python
df1 = pd.merge(df_ord, df_cust, on = 'Customer ID',how = 'left')
```

```python
df_all = pd.merge(df1, df_prod, on = 'Product ID',how = 'left')
```

```python
df = df_all.copy()
```

```python
df['Year'] = df['Order Date'].dt.year
df['Month'] = df['Order Date'].dt.month
df['Day'] = df['Order Date'].dt.day
df['WeekDay'] = df['Order Date'].dt.day_name()
```
